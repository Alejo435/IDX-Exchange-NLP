## 1. Dataset Overview

```python
# column types and missing values across the sample
overview = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "non_null": df.notna().sum(),
    "null_pct": (df.isna().mean() * 100).round(1),
})
overview
```
